## Exercise 1: Sigmoid Function


Implement the sigmoid function and plot it.

````python
import matplotlib.pyplot as plt
# Implement sigmoid: σ(z) = 1 / (1 + e^{-z})
# Plot it for z in range [-10, 10]
# Verify: sigmoid(0) = 0.5, sigmoid(large) ≈ 1, sigmoid(very negative) ≈ 0

# Your code here:

def sigmoid(z):
    return (1 / (1 + np.exp(-z)))

x = [i for i in range(-10, 10)]
y = [sigmoid(z) for z in x]


plt.plot(x, y, marker='o', linestyle='-', color='blue', label='My Data')

# 3. Add labels and title
plt.xlabel('z')
plt.ylabel('σ(z)')
plt.title('Sigmoid')

# 4. Display the chart
plt.legend()
plt.show()


````

## Exercise 2: Binary Cross-Entropy Loss

Implement BCE loss and compute it for example predictions.

````python
# BCE: L = -(1/n) * sum(y * log(y_hat) + (1 - y) * log(1 - y_hat))
#

# Your code here:
def bce(y, y_hat):
    return -(1/2) * (y * np.log(y_hat) + (1 - y) * np.log(1- y_hat))


# Test cases:
#   y=1, y_hat=0.99 → loss should be very small (confident and correct)
print(f"y=1, y_hat=0.99: {bce(1, .99)}")
#   y=1, y_hat=0.01 → loss should be very large (confident and wrong)
print(f"y=1, y_hat=0.01: {bce(1, .01)}")
#   y=0, y_hat=0.01 → loss should be very small
print(f"y=0, y_hat=0.01: {bce(0, .01)}")
#   y=0, y_hat=0.99 → loss should be very large
print(f"y=0, y_hat=0.99: {bce(0, .99)}")
````

## Exercise 3: Logistic Regression from Scratch

Implement logistic regression with gradient descent using BCE loss.

````python
# Generate 2D data: two Gaussian blobs
np.random.seed(42)
n = 200
X_class0 = np.random.randn(n // 2, 2) + np.array([-1, -1])
X_class1 = np.random.randn(n // 2, 2) + np.array([1, 1])
X_clf = np.vstack([X_class0, X_class1])
y_clf = np.concatenate([np.zeros(n // 2), np.ones(n // 2)]).reshape(-1, 1)
print(f"X_clf.shape: {X_clf.shape}")
print(f"y_clf.shape: {y_clf.shape}")
print(X_clf)
````

````python
# Implement logistic regression:
#   y_hat = sigmoid(X @ w + b)
#   Gradient: dL/dw = (1/n) * X.T @ (y_hat - y)
#             dL/db = (1/n) * sum(y_hat - y)
#
# Train on X_clf, y_clf. Print final weights, bias, and loss.

# Your code here:
def logistic_regression(X, y, w, b, iterations= 10, learning_rate=.01):
    n = X.shape[0]

    bce_list = []
    
    for i in range(iterations):
        # compute the gradient 
        y_hat = sigmoid(X @ w + b)

        # compute the MSE
        bce_list.append(np.mean(bce(y, y_hat)))

        dL_dw = (1/n) * X.T @ (y_hat - y)
        dL_db = (1/n) * sum(y_hat - y)
        #print(f"y.shape: {y.shape}")
        #print(f"y_hat.shape: {y_hat.shape}")
        #print(f"dL_dw.shape: {dL_dw.shape}")
        #print(f"dL_db.shape: {dL_db.shape}")
        w -= (learning_rate * dL_dw)
        b -= (learning_rate * dL_db)
    
    return w, b, bce_list

w = np.ones((2, 1))
b = 1
epochs = 1000
alpha = .01


w, b, bce_list =  logistic_regression(X_clf, y_clf, w, b, iterations=epochs, learning_rate=alpha)
print(f"w: {w}")
print(f"b: {b}")
print(f"Loss: {bce_list[len(bce_list) - 1]}")
````

## Exercise 5: Imbalanced Dataset — Precision, Recall, F1

Create an imbalanced dataset (95% class 0, 5% class 1). Train a model and show why accuracy is misleading.

````python
# Generate imbalanced data
np.random.seed(42)
n_majority = 950
n_minority = 50
X_maj = np.random.randn(n_majority, 2) + np.array([0, 0])
X_min = np.random.randn(n_minority, 2) + np.array([2, 2])
X_imb = np.vstack([X_maj, X_min])
y_imb = np.concatenate([np.zeros(n_majority), np.ones(n_minority)]).reshape(-1, 1)
````

````python
# Train logistic regression on this data.
# Compute predictions (threshold at 0.5).
# Then compute and print:
#   - Accuracy
#   - Confusion matrix (TP, FP, TN, FN)
#   - Precision = TP / (TP + FP)
#   - Recall = TP / (TP + FN)
#   - F1 = 2 * (precision * recall) / (precision + recall)
#
# Compare: what would accuracy be if you just predicted 0 for everything?
# Your code here:
w = np.ones((2, 1))
b = 1
epochs = 1000
alpha = .01


w, b, bce_list =  logistic_regression(X_imb, y_imb, w, b, iterations=epochs, learning_rate=alpha)
print(f"w: {w}")
print(f"b: {b}")
print(f"Loss: {bce_list[len(bce_list) - 1]}")

# TP (True Positive): Correctly predicts 1
# FP (False Positive): Incorrectly predicts 1
# TN (True Negative): Correctly predicts 0
# FN (False Negative): Incorrectly predicts 0

# make the predictions
y_hat = sigmoid(X_imb @ w + b)                                                                                                                                 
predictions = (y_hat >= 0.5).astype(int)

# compare the predictions with the actual values (y_imb)
tp = ((predictions == 1) & (y_imb == 1)).sum()
fp = ((predictions == 1) & (y_imb == 0)).sum()
tn = ((predictions == 0) & (y_imb == 0)).sum()
fn = ((predictions == 0) & (y_imb == 1)).sum()

# build the confusion matrix
confusion_matrix = np.array([[tp, fp], [tn, fn]])

# compute the accuracy
accuracy = (tp + tn) / (tp + fp + tn + fn)
# compute the precision
precision = tp / (tp + fp)
# compute the recall
recall = tp / (tp + fn)
# compute F1 (Harmonic Mean)
F1 = 2 * (precision * recall) / (precision + recall)
print(f"Confusion Matrix: {confusion_matrix}")
print(f"Accuracy: {accuracy}")
print(f"Precision: {precision}")
print(f"Recall: {recall}")
print(f"F1: {F1}")

# If I had just predicted zero for everything, the accuracy would have been 95% due to the imbalanced dataset
````

## Exercise 6: ROC Curve and AUC

Plot the ROC curve and compute AUC from scratch.

````python
# Using your trained model's raw probabilities (before thresholding):
# 1. Vary the threshold from 0 to 1 (e.g., 100 values with np.linspace)
# 2. At each threshold, compute TPR (recall) and FPR = FP / (FP + TN)
# 3. Plot FPR (x-axis) vs TPR (y-axis)
# 4. Compute AUC using the trapezoidal rule: np.trapz(tpr_list, fpr_list)
# 5. Plot the diagonal (random classifier) for reference

# Your code here:

# this is our raw data: X_clf and y_clf

# threshold range
thresholds = np.linspace(0, 1, 100) # 100 values from 0 to 1
                                                                                                                         
tpr_list = []                                                                                                                                                  
fpr_list = []                                                                                                                                                  
                                                                                                                                                             
for t in thresholds:                                                                                                                                           
    preds = (y_hat >= t).astype(int)                                                                                                                           
    tp = ((preds == 1) & (y_imb == 1)).sum()                                                                                                                   
    fp = ((preds == 1) & (y_imb == 0)).sum()
    fn = ((preds == 0) & (y_imb == 1)).sum()                                                                                                                   
    tn = ((preds == 0) & (y_imb == 0)).sum()                                                                                                                 
    tpr_list.append(tp / (tp + fn))
    fpr_list.append(fp / (fp + tn))
    
# FPR = FP / (FP + TN)
# TPR = TP / (TP + FN) where TP + FN is the total number of actual 1's

x = np.linspace(0, 1, 100)
plt.plot(x, x, label="Random Classifier", color="red", linewidth=2, linestyle=':')


plt.plot(fpr_list, tpr_list, label="ROC")
plt.title("ROC Curve")
plt.xlabel("FPR") # False Positive Rate (FPR): proportion of negative cases incorrectly identified as positive by the model
plt.ylabel("TPR") # True Positive Rate (TPR): proportion of postive cases correctly identified as positve by the model 
plt.legend() 

plt.show()

auc = np.trapezoid(tpr_list[::-1], fpr_list[::-1])
print(f"AUC: {auc}")
````

## Exercise 7: Softmax and Multiclass Classification

Implement softmax and categorical cross-entropy. Train on a 3-class dataset.

````python
# Generate 3-class data
np.random.seed(42)
n_per_class = 100
X_c0 = np.random.randn(n_per_class, 2) + np.array([-2, 0])
X_c1 = np.random.randn(n_per_class, 2) + np.array([2, 0])
X_c2 = np.random.randn(n_per_class, 2) + np.array([0, 3])
X_multi = np.vstack([X_c0, X_c1, X_c2])
y_multi = np.concatenate([np.zeros(n_per_class), np.ones(n_per_class), 2 * np.ones(n_per_class)]).astype(int)
````

````python
# 1. Implement softmax: softmax(z)_i = exp(z_i) / sum(exp(z_j))
#    Hint: subtract max(z) before exp for numerical stability
def softmax(z):                                                                                                                                                
    z_shifted = z - np.max(z, axis=1, keepdims=True)
    exp_z = np.exp(z_shifted)                                                                                                                                  
    return exp_z / np.sum(exp_z, axis=1, keepdims=True)
# 2. One-hot encode y_multi: shape (300,) → (300, 3)
#    Hint: np.eye(3)[y_multi]
y_encoded = np.eye(3)[y_multi]
# 3. Implement categorical cross-entropy:
#    L = -(1/n) * sum(Y_onehot * log(Y_hat))
def cat_cross_entropy(Y_onehot, Y_hat):
    n = Y_onehot.shape[0]
    L = -1 * (1 / n) * sum(Y_onehot * np.log(Y_hat))
    return L
    
# 4. Train with gradient descent:
#    W is (2, 3), b is (1, 3)
#    y_hat = softmax(X @ W + b)
#    dL/dW = (1/n) * X.T @ (y_hat - Y_onehot)
#    dL/db = (1/n) * sum(y_hat - Y_onehot, axis=0)

def gradient_descent(X, Y_onehot, w, b, iterations= 1000, learning_rate=.01):
    n = X.shape[0]
    for i in range(iterations):
        # compute the gradient 
        y_hat = softmax(X @ w + b)
        dL_dw = (1/n) * X.T @ (y_hat - Y_onehot)
        dL_db = (1/n) * np.sum(y_hat - Y_onehot, axis=0)

        w -= (learning_rate * dL_dw)
        b -= (learning_rate * dL_db)

    return w, b

w = np.ones((2, 3))
b = np.ones((1, 3))

w, b = gradient_descent(X_multi, y_encoded, w, b)
print(f"w: {w}")
print(f"b: {b}")

````

````python
# 9c: FP32 vs FP16 comparison
# 1. Create a (1000, 1000) matrix in float32 and float64. Compare memory usage:
#    print(matrix.nbytes) for each
# 2. Multiply them (@ operator). Time both with %%timeit or time.time().
#    FP32 should use half the memory. Speed difference depends on hardware.
# 3. Compute the MSE between the FP32 and FP64 results of the same multiplication.
#    This shows the precision you lose going from 64 to 32 bit.
# 4. Repeat with float16. How much precision is lost? How much memory saved?
import time
# Your code here:
# 1.
n = 1000
matrix_flt32 = np.random.rand(n, n).astype(np.float32)
matrix_flt64 = np.random.rand(n, n) # defaults to float64
print(f"matrix_flt32.nbytes: {matrix_flt32.nbytes}")
print(f"matrix_flt64.nbytes: {matrix_flt64.nbytes}")
# flt64 uses double the memory
# 2.
start = time.time()
result_flt32 = matrix_flt32 @ matrix_flt32

end = time.time()
print(f"Time (float32): {(end - start):.2}")

start = time.time()
result_flt64 = matrix_flt64 @ matrix_flt64

end = time.time()
print(f"Time (float64): {(end - start):.2}")

# 3.
mse = (1 / n) * np.mean((result_flt32 - result_flt64)**2)
print(f"mse(flt64 and flt32): {mse:.5f}")
# 4.
matrix_flt16 = np.random.rand(n, n).astype(np.float16)
print(f"matrix_flt16.nbytes: {matrix_flt16.nbytes}") # 1/4 the memory of float64
start = time.time()
result_flt16 = matrix_flt16 @ matrix_flt16

end = time.time()
print(f"Time (float16): {(end - start):.2}")
mse = (1 / n) * np.mean((result_flt16 - result_flt64)**2)
print(f"mse(flt64 and flt16): {mse:.5f}")



# on my machine I don't really see any performance gain for matrix multiplication
# with less precision.
````
