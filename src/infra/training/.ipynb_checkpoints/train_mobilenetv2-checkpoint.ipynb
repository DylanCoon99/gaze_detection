# MobileNetV2 Training — Full 300W-LP → AFLW2000 Eval

Train a frozen MobileNetV2 backbone with regression head on the full 300W-LP dataset.
Test on AFLW2000 (standard benchmark split).

```python
import os
os.environ["KAGGLEHUB_CACHE"] = "/Volumes/T7/kaggle_datasets"

import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import scipy.io as sio
import kagglehub
from torchvision import transforms, models
from torch.utils.data import Dataset, DataLoader, ConcatDataset
from pathlib import Path
import torchvision.io as tv_io
```

## Download Datasets

```python
# Download 300W-LP (train) — already cached if previously downloaded
train_path = kagglehub.dataset_download("maulidio16/300w-lp")
print(f"300W-LP path: {train_path}")

# Download AFLW2000 (test)
test_path = kagglehub.dataset_download("mohamedadlyi/aflw2000-3d")
print(f"AFLW2000 path: {test_path}")
```

```python
# Training data paths — all 300W-LP folders (no flips)
TRAIN_DIRS = [
    os.path.join(train_path, "300W_LP", "AFW"),
    os.path.join(train_path, "300W_LP", "HELEN"),
    os.path.join(train_path, "300W_LP", "LFPW"),
    os.path.join(train_path, "300W_LP", "IBUG"),
]

# Test data path — AFLW2000
# Check the actual folder structure after download
for d in TRAIN_DIRS:
    count = len(list(Path(d).glob("*.mat")))
    print(f"{Path(d).name}: {count} samples")

print(f"\nTest path contents: {os.listdir(test_path)}")
```

## Dataset Class

```python
class GazeDataset(Dataset):

    def __init__(self, data_path, transform=None):
        self.data_path = data_path
        self.transform = transform
        self.samples = []

        path = Path(data_path)
        for mat_file in sorted(path.glob("*.mat")):
            img_file = mat_file.with_suffix(".jpg")
            if not img_file.exists():
                continue

            mat = sio.loadmat(mat_file)
            pose = mat['Pose_Para'][0]
            pitch, yaw = pose[0], pose[1]
            pitch = pitch * 180 / np.pi
            yaw = yaw * 180 / np.pi

            label = torch.tensor([yaw, pitch], dtype=torch.float32)
            self.samples.append((img_file, label))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        img_path, label = self.samples[idx]
        image_tensor = tv_io.read_image(str(img_path))
        if self.transform:
            image_tensor = self.transform(image_tensor)
        return (image_tensor, label)
```

## Create DataLoaders

```python
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ConvertImageDtype(torch.float32),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

# Combine all 300W-LP folders into one training set
train_datasets = [GazeDataset(d, transform=transform) for d in TRAIN_DIRS]
train_dataset = ConcatDataset(train_datasets)

# AFLW2000 test set — update path after checking folder structure above
# You may need to adjust this path based on what the download cell prints
TEST_DIR = os.path.join(test_path, "AFLW2000")
test_dataset = GazeDataset(TEST_DIR, transform=transform)

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True, num_workers=0)
test_loader = DataLoader(test_dataset, batch_size=32)

print(f"Training samples: {len(train_dataset)}")
print(f"Test samples:     {len(test_dataset)}")
```

## Build Model — Frozen MobileNetV2 + Regression Head

```python
# Device setup
if torch.backends.mps.is_available():
    device = torch.device("mps")
    print("MPS device found!")
else:
    device = torch.device("cpu")
    print("MPS not available. Using CPU.")

# Load pretrained MobileNetV2 and freeze backbone
model = models.mobilenet_v2(weights='IMAGENET1K_V1')

for param in model.parameters():
    param.requires_grad = False

# Replace classifier with regression head
model.classifier = nn.Sequential(
    nn.Dropout(0.2),
    nn.Linear(1280, 2)
)

model.to(device)
print(f"Trainable params: {sum(p.numel() for p in model.parameters() if p.requires_grad)}")
print(f"Total params:     {sum(p.numel() for p in model.parameters())}")
```

## Training Loop

```python
EPOCHS = 15
loss_fn = nn.MSELoss()
optimizer = optim.Adam(model.classifier.parameters(), lr=0.001)

model.train()

for epoch in range(EPOCHS):
    total_mse = 0.0
    total_mae = 0.0
    num_samples = 0

    for i, (images, labels) in enumerate(train_loader):
        images = images.to(device)
        labels = labels.to(device)

        y_hat = model(images)

        loss = loss_fn(y_hat, labels)
        total_mse += loss.item() * labels.size(0)
        total_mae += torch.abs(y_hat - labels).detach().sum().item()
        num_samples += labels.size(0)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        if i % 200 == 0:
            print(f"  Epoch {epoch}, batch {i}/{len(train_loader)}, loss: {loss.item():.4f}")

    avg_mse = total_mse / num_samples
    avg_mae = total_mae / (num_samples * 2)  # 2 outputs per sample
    print(f"Epoch {epoch} — MSE: {avg_mse:.4f}, MAE: {avg_mae:.2f}°")

# Save immediately after training
model.cpu()
model.eval()
scripted = torch.jit.trace(model, torch.randn(1, 3, 224, 224))
torch.jit.save(scripted, "/Volumes/T7/models/mobilenet_v2_fp32.pt")
print("\nModel saved to /Volumes/T7/models/mobilenet_v2_fp32.pt")
```

## Evaluate on AFLW2000 Test Set

```python
import os                                                                                                                                                      
for root, dirs, files in os.walk(test_path):                                                                                                                   
    level = root.replace(test_path, '').count(os.sep)                                                                                                          
    if level < 2:                                                                                                                                              
        print(f"{root}: {len(files)} files, subdirs: {dirs}")
```

```python
model.to(device)                            
model.eval()

total_mae_yaw = 0.0
total_mae_pitch = 0.0
num_samples = 0

with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)
        labels = labels.to(device)

        y_hat = model(images)
        errors = torch.abs(y_hat - labels)

        total_mae_yaw += errors[:, 0].sum().item()
        total_mae_pitch += errors[:, 1].sum().item()
        num_samples += labels.size(0)

mae_yaw = total_mae_yaw / num_samples
mae_pitch = total_mae_pitch / num_samples
mae_combined = (mae_yaw + mae_pitch) / 2

print(f"\nAFLW2000 Test Results:")
print(f"  MAE (yaw):      {mae_yaw:.2f}°")
print(f"  MAE (pitch):    {mae_pitch:.2f}°")
print(f"  MAE (combined): {mae_combined:.2f}°")
```

## Fine-tune — Unfreeze Backbone

```python
# Unfreeze all parameters
for param in model.parameters():
    param.requires_grad = True

# Lower learning rate for fine-tuning
model.to(device)
model.train()

EPOCHS_FT = 10
optimizer_ft = optim.Adam(model.parameters(), lr=0.0001)

for epoch in range(EPOCHS_FT):
    total_mse = 0.0
    total_mae = 0.0
    num_samples = 0

    for i, (images, labels) in enumerate(train_loader):
        images = images.to(device)
        labels = labels.to(device)

        y_hat = model(images)

        loss = loss_fn(y_hat, labels)
        total_mse += loss.item() * labels.size(0)
        total_mae += torch.abs(y_hat - labels).detach().sum().item()
        num_samples += labels.size(0)

        optimizer_ft.zero_grad()
        loss.backward()
        optimizer_ft.step()

        if i % 200 == 0:
            print(f"  Epoch {epoch}, batch {i}/{len(train_loader)}, loss: {loss.item():.4f}")

    avg_mse = total_mse / num_samples
    avg_mae = total_mae / (num_samples * 2)
    print(f"Epoch {epoch} — MSE: {avg_mse:.4f}, MAE: {avg_mae:.2f}°")

# Save fine-tuned model immediately
model.cpu()
model.eval()
scripted = torch.jit.trace(model, torch.randn(1, 3, 224, 224))
torch.jit.save(scripted, "/Volumes/T7/models/mobilenet_v2_fp32_finetuned.pt")
print("\nFine-tuned model saved to /Volumes/T7/models/mobilenet_v2_fp32_finetuned.pt")
```
